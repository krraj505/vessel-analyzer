```python
import pandas as pd
from prophet import Prophet

# 1. Load Member B's cleaned CMO dataset
df_clean = pd.read_csv('cleaned_cmo_monthly_prices.csv')

# 2. Automatically map Date and Price columns
date_col = [c for c in df_clean.columns if any(k in c.lower() for k in ['date', 'month', 'ds'])][0]
price_col = [c for c in df_clean.columns if c != date_col][0]

df_prophet = pd.DataFrame({
    'ds': pd.to_datetime(df_clean[date_col]),
    'y': pd.to_numeric(df_clean[price_col], errors='coerce')
}).dropna().sort_values('ds')

# 3. Train Prophet Model
model = Prophet(
    yearly_seasonality=True,
    weekly_seasonality=False,
    daily_seasonality=False,
    changepoint_prior_scale=0.05
)
model.fit(df_prophet)

# 4. Generate 30-Day Forecast & Export Clean JSON
future = model.make_future_dataframe(periods=30, freq='D')
forecast = model.predict(future)

forecast_30day = forecast[['ds', 'yhat', 'yhat_lower', 'yhat_upper']].tail(30).copy()
forecast_30day['ds'] = forecast_30day['ds'].dt.strftime('%Y-%m-%d')
forecast_30day = forecast_30day.round(2)

forecast_30day.to_json('forecast_30day.json', orient='records', indent=4)
print("30-day forecast successfully updated!")
```
